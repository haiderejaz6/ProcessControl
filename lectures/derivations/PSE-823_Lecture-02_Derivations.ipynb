## Part A -- Your Turn, full derivation (Problem 2.9 setup)

Original steady state: $v_1=10$, $v_2=20$ L/min, $C_a(0)=3$ g/L, $V=150$ L.

At 3:00 PM, stream 1 increases to $20$ L/min; stream 2 and stream 1's concentration unchanged.

**New total flow:** $v_3=v_1+v_2=20+20=40$ L/min (different from the book's own scenario, where $v_3$ stayed at 30 throughout).

**New steady-state feed concentration:**
$$C_{a3,new}=\frac{(20)(1)+(20)(4)}{40}=\frac{100}{40}=2.5\ \text{g/L}$$

**New residence time:**
$$\tau_{new}=\frac{V}{v_3}=\frac{150}{40}=3.75\ \text{min}$$

**ODE and transform:**
$$3.75\frac{dC_a}{dt}+C_a=2.5,\quad C_a(0)=3$$
$$C_a(s)=\frac{2.5}{s(3.75s+1)}+\frac{11.25}{3.75s+1}$$

```python
import sympy as sp

t, s = sp.symbols('t s', positive=True)
Ca = sp.Function('C_a')
ode = sp.Eq(sp.Rational(15, 4) * Ca(t).diff(t) + Ca(t), sp.Rational(5, 2))
sol = sp.dsolve(ode, Ca(t), ics={Ca(0): 3})
print(sol)

Ca_s = sp.laplace_transform(sol.rhs, t, s, noconds=True)
print(sp.nsimplify(sp.simplify(Ca_s)))
```

## Part B -- Your Turn, full derivation (Problem 2.9 inversion)

Convert to standard $(s+a)$ form: $3.75s+1=3.75(s+0.2\overline{6})$, where $0.2\overline{6}=1/3.75=4/15$.

**First term:**
$$\frac{2.5}{s(3.75s+1)}=\frac{0.6\overline{6}}{s(s+0.2\overline{6})}=\frac{A}{s}+\frac{B}{s+0.2\overline{6}}$$
Cover-up: $A=2.5$, $B=-2.5$

**Second term:** $\frac{11.25}{3.75(s+0.2\overline{6})}=\frac{3}{s+0.2\overline{6}}$

**Sum:**
$$C_a(s)=\frac{2.5}{s}+\frac{0.5}{s+0.2\overline{6}} \;\Rightarrow\; C_a(t)=2.5+0.5e^{-t/3.75}\ \text{g/L}$$

**Self-check:** $t=0$: $2.5+0.5=3$ (matches IC). $t\to\infty$: $C_a\to2.5$ (matches Part A's new steady state).

```python
import sympy as sp

s, t = sp.symbols('s t', positive=True)
Ca_s = sp.Rational(5, 2) / (s * (sp.Rational(15, 4) * s + 1)) \
       + sp.Rational(45, 4) / (sp.Rational(15, 4) * s + 1)
print(sp.apart(Ca_s))
print(sp.inverse_laplace_transform(Ca_s, s, t))
```

## Reference -- the book's own worked examples, full detail

**Example 3.1 (basic):** $x(s)=\frac{1}{s(s+1)}=\frac{1}{s}-\frac{1}{s+1} \Rightarrow x(t)=1-e^{-t}$

**Example 3.4 (complex roots):** $x(s)=\frac{2}{s(s^2+2s+2)}$. $x(s)=\frac{1}{s}-\frac{s+2}{s^2+2s+2}$. Complete the square: $s^2+2s+2=(s+1)^2+1$. Regroup: $x(s)=\frac{1}{s}-\frac{(s+1)+1}{(s+1)^2+1^2}$. Invert with $\mathcal{L}\{e^{-at}\cos kt\}=\frac{s+a}{(s+a)^2+k^2}$, $\mathcal{L}\{e^{-at}\sin kt\}=\frac{k}{(s+a)^2+k^2}$: $x(t)=1-e^{-t}(\cos t+\sin t)$.

**Example 3.5 (repeated roots):** $x(s)=\frac{1}{s(s+1)^3}$. Setup: $x(s)=\frac{A}{s}+\frac{B}{(s+1)^3}+\frac{C}{(s+1)^2}+\frac{D}{s+1}$. Cover-up: $A=1$, $B=-1$. Matching coefficients: $1+D=0$, $3+C+2D=0$, $2+C+D=0 \Rightarrow D=-1$, $C=-1$. Final: $x(t)=1-e^{-t}(\frac{t^2}{2}+t+1)$.
