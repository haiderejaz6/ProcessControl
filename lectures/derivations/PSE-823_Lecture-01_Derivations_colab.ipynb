### Gain comparison (System A vs. System B)

$$e_{ss} = \frac{\Delta d}{1+K_cK_p}$$

Shows $K_c\to\infty$ drives offset toward zero but never reaches it -- motivating integral action. Instability needs the *dynamic* transfer function (with measurement lag), not just steady-state gain. Today is vocabulary and the warning; the algebra for *why* instability happens is coming.

```python
import sympy as sp

Kc, Kp, d, e = sp.symbols('K_c K_p Delta_d e_ss', positive=True)
eq = sp.Eq(e, d / (1 + Kc * Kp))
sol = sp.solve(eq.subs(e, 0), Kc)
print("Finite Kc solving e_ss=0:", sol)  # empty -> confirms it's a limit, not attainable
```
